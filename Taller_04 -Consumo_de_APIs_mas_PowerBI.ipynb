```python
# Taller 04 - Consumo de API y Exportacion a CSV
# Estudiante: Pablo Montenegro

import requests
import pandas as pd

# 1. PASO: Guardar la direccion de la API que vamos a consultar
# Usamos CoinGecko para traer las principales 50 criptomonedas en dolares
url_api = "https://api.coingecko.com/api/v3/coins/markets?vs_currency=usd&order=market_cap_desc&per_page=50&page=1"

# 2. PASO: Hacer la peticion a la API con la libreria requests
respuesta = requests.get(url_api)

# Convertimos la respuesta de la API (formato JSON) en una lista de Python
datos_json = respuesta.json()


# 3. PASO: Pasamos los datos a una tabla de Pandas (DataFrame)
df_criptos = pd.DataFrame(datos_json)


# 4. PASO: Seleccionamos solo las columnas que necesitamos para el reporte
columnas_guardar = ['name', 'symbol', 'current_price', 'market_cap', 'total_volume', 'price_change_percentage_24h']
df_filtrado = df_criptos[columnas_guardar]


# 5. PASO: Renombramos las columnas a español para que se entiendan facil
df_filtrado = df_filtrado.rename(columns={
    'name': 'Nombre',
    'symbol': 'Simbolo',
    'current_price': 'Precio_USD',
    'market_cap': 'Capitalizacion',
    'total_volume': 'Volumen_24h',
    'price_change_percentage_24h': 'Cambio_24h_Porcentaje'
})


# 6. PASO: Redondeamos los decimales para que los numeros queden limpios
df_filtrado['Precio_USD'] = df_filtrado['Precio_USD'].round(2)
df_filtrado['Cambio_24h_Porcentaje'] = df_filtrado['Cambio_24h_Porcentaje'].round(2)


# 7. PASO: Guardamos el resultado en un archivo CSV listo para Power BI
df_filtrado.to_csv('datos_cripto_powerbi.csv', index=False)

print("¡Listo! Se creo el archivo datos_cripto_powerbi.csv con exito.")
```
